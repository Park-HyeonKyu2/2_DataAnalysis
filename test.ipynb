```python
class HashTable:
    def __init__(self, size=10):
        self.size = size
        # 충돌을 고려하지 않으므로 빈 칸(None)으로 초기화
        self.table = [None] * self.size

    def _hash_function(self, key):
        """키를 테이블 크기로 나눈 나머지를 인덱스로 반환"""
        return hash(key) % self.size

    def put(self, key, value):
        """데이터 삽입 및 수정 (충돌 나면 기존 데이터는 증발)"""
        index = self._hash_function(key)
        self.table[index] = (key, value)

    def get(self, key):
        """데이터 검색"""
        index = self._hash_function(key)
        pair = self.table[index]
        
        # 방이 비어있지 않고, 저장된 키가 내가 찾는 키와 일치할 때만 반환
        if pair is not None and pair[0] == key:
            return pair[1]
        return None

ht = HashTable()

# 데이터 삽입
ht.put("Apple", "사과")
ht.put("Banana", "바나나")

# 데이터 가져오기
print(ht.get("Apple"))   # 출력: 사과
print(ht.get("Banana"))  # 출력: 바나나

# 데이터 삭제
ht.delete("Apple")
print(ht.get("Apple"))   # 출력: None
```

```python
'''
1번과 2번비교하고 큰 수는 교환 -> 2번과 3번비교
올라가다가 마지막까지 비교하고 마지막은 가장 큰 수가 배치
그러므로 반복을 돌 때마다 리스트 길이 - 1 해줘야 한다.
'''
import random as r

def bubble_asc(arr):
    for i in range(len(arr)-1):
        for j in range(len(arr)-i - 1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
            else:
                pass
    
    print(arr)

def bubble_desc(arr):
    for i in range(len(arr)-1):
        for j in range(len(arr)-i - 1):
            if arr[j] < arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
            else:
                pass
    
    print(arr)

datas = list(range(10))
r.shuffle(datas)
print(datas[:])
bubble_asc(datas)
bubble_desc(datas)
```

```python
'''
가장 처음의 값을 min_num이라 설정
차례로 앞의 값들과 비교하여 작은 수가 있다면 min_num에 값과 인덱스 복사
마지막 까지 탐색하였다면 min_num의 값을 가장 앞에 삽입
'''
import random as r

def selection_asc(arr):
    for i in range(len(arr)-1):
        min_num = [i, arr[i]]
        for j in range(i + 1, len(arr)):
            if min_num[1] > arr[j]:
                min_num[0], min_num[1] = j, arr[j]
            else:
                pass
        arr[i], arr[min_num[0]] = arr[min_num[0]], arr[i]
    print(arr)

def selection_desc(arr):
    for i in range(len(arr)-1):
        max_num = [i, arr[i]]
        for j in range(i + 1, len(arr)):
            if max_num[1] < arr[j]:
                max_num[0], max_num[1] = j, arr[j]
            else:
                pass
        arr[i], arr[max_num[0]] = arr[max_num[0]], arr[i]
    print(arr)


datas = list(range(10))
r.shuffle(datas)
print(datas)
selection_asc(datas[:])
selection_desc(datas[:])
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def googlescholar(keyword, max_page=5):
    headers = {
        "User-Agent": "Mozilla/5.0",
        "Referer": "https://scholar.google.com/?hl=ko"
    }

    all_data = []

    for page in range(0, max_page):
        time.sleep(3)
        start = page * 10
        url = (
            f"https://scholar.google.com/scholar?start={start}&q={keyword}&hl=ko&as_sdt=0,5"
        )
        
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        rows = soup.select("div#gs_res_ccl_mid div.gs_r.gs_or.gs_scl")

        for row in rows:
            title_tag = row.select_one("h3.gs_rt a")

            all_data.append({
                "제목": title_tag.get_text(strip=True)
            })
        time.sleep(1)
        
    df = pd.DataFrame(all_data)
    return df

googlescholar("learning", 10)
```
